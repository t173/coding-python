```python
2 + 3 + 7
```

Basic arithmetic operators are

  | Operator | Meaning |
  | :--: | :--- |
  | `+` | addition |
  | `-` | subtraction or negation |

### Division

In Python, there are two division operators, `/` and `//`.  The single slash
`/` is for floating point division; the result is always a `float` and contains and the double slash `//` is integer division.  Integer division is what you may have learned in elementary school.  To divide $a$ by $b$, we wish to find the quotient $q$ and remainder $r$ such that $$a = bq + r$$

The quotient $q$ is `a//b` and the remainder $r$ is `a % b`, where `%` is called the modulo operator.  Both results are integers.

```python
print("Floating point division:", 8/3, type(8/3))
print("Integer division, quotient:", 8//3, type(8//3))
print("Integer division, remainder:", 8 % 3, type(8 % 3))
```
