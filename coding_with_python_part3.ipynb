Let's first open the file and read the first line:

```python
f = open('crosswd.txt')
f.readline()
```

The `open()` function takes the name of the file you want to open and returns a file handler, which is an object we use to access the file.

The `readline()` method of the file handler reads the next line of the file and returns it as a string.

This crossword list is sorted alphabetically.  The first word is ['A'ā](https://en.wikipedia.org/wiki/Lava#%CA%BBA%CA%BB%C4%81), which comes from Hawaiian and refers to a specific type of lava flow.  For simplicity, it is written without diacritics or capitalization: `aa`.

The `\n` at the end is a **newline character**, which is used to separate lines in text.  The `readline()` method reads until the next newline character (or the end of the file).

Using a `for` loop, we can read several more lines of text:

```python
for i in range(5):
  print(f.readline())
```

The file handler `f` maintains a cursor position within the file, so the lines printed are those after the line containing `aa\n`.  Notice when using
`print()`, just as the quotation marks around a string are not printed, newline characters are not shown as `\n`, but actually create a new line.  Since
`print()` normally creates a new line at the end of what it prints, two new lines are printed per line; one from the `\n` in the string and one from the `print()` function.

### Stripping whitespace

We can remove the newline character at the end of each string we read using the string method `strip()`, which strips whitespace (spaces, newlines, etc.) from the beginning and end of the string.

```python
for i in range(5):
  print(f.readline().strip())
```

Notice how the printing is no longer double spaced.

When you are finished with the file, close it:

```python
f.close()
```
